```python
import pandas as pd
data = {
    "OrderID": [101,102,103,104,105,106,107,108,109,110,
                111,112,113,114,115,116,117,118,119,120],

    "Product": ["Laptop","Phone","Tablet","Laptop","Headphones",
                "Phone","Laptop","Tablet","Phone","Headphones",
                "Laptop","Phone","Tablet","Headphones","Laptop",
                "Phone","Tablet","Laptop","Headphones","Phone"],

    "Category": ["Electronics","Electronics","Electronics","Electronics",
                 "Accessories","Electronics","Electronics","Electronics",
                 "Electronics","Accessories","Electronics","Electronics",
                 "Electronics","Accessories","Electronics","Electronics",
                 "Electronics","Electronics","Accessories","Electronics"],

    "City": ["Kolkata","Delhi","Mumbai","Kolkata","Delhi",
             "Mumbai","Delhi","Kolkata","Mumbai","Delhi",
             "Kolkata","Delhi","Mumbai","Kolkata","Delhi",
             "Mumbai","Kolkata","Delhi","Mumbai","Kolkata"],

    "Gender": ["F","M","F","M","F","M","M","F","F","M",
               "F","M","F","M","F","M","F","M","F","M"],

    "Age": [25,32,28,41,23,35,29,26,31,24,
            38,27,22,34,30,36,25,42,28,33],

    "Quantity": [1,2,1,1,3,2,1,2,1,2,
                 1,3,2,1,1,2,3,1,2,1],

    "Price": [60000,45000,30000,65000,3000,
              48000,62000,28000,47000,3500,
              61000,44000,32000,4000,63000,
              46000,29000,68000,3200,49000]
}
df = pd.DataFrame(data)
df
```

```python
# Which city generated the highest total Sales?
A=df.groupby("City")["Sales"].sum().reset_index()
A.loc[A["Sales"].idxmax(),"City"]
A.sort_values("Sales",ascending=False).loc[0,"City"]
```

```python
# Which product generated the highest total Sales?
A=df.groupby("Product")["Sales"].sum().reset_index()
A.sort_values("Sales",ascending=False).loc[0,"Product"]
```

```python
# Create a bar plot of total sales by product
import matplotlib.pyplot as plt
A=df.groupby("Product")["Sales"].agg("sum")
plt.bar(A.index,A.values)
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

```python
# Create a count plot of:Orders by City
import seaborn as sns
sns.countplot(df,x="City")
plt.ylabel("Counts")
plt.xlabel("City")
plt.show()
```

```python
# Create a box plot of:Sales Look at whether there are potential outliers.
A=plt.boxplot([df[df["Gender"]=="F"]["Sales"],df[df["Gender"]=="M"]["Sales"]])
plt.xticks([1,2],["Male","Female"])
plt.ylabel("Sales")
plt.show()
df.boxplot(column="Sales",by="Gender")
plt.xlabel("Gender")
plt.ylabel("Sales")
plt.show()
sns.boxplot(df,y="Sales",hue="Gender")
plt.ylabel("Sales")
plt.xlabel("Gender")
plt.show()
```

```python
# Create a scatter plot of:Age vs Sales
plt.scatter(df["Age"],df["Sales"])
plt.xlabel("Age")
plt.ylabel("Sales")
plt.show()
df.plot.scatter(x="Age",y="Sales")
plt.show()
sns.scatterplot(data=df,x="Age",y="Sales")
plt.show()
```

```python
# Add a regression/trend line to the Age vs Sales relationship.
import numpy as np
x=df["Age"]
y=df["Sales"]
plt.scatter(x,y)
m,c=np.polyfit(x,y,1)
plt.plot(x,m*x+c)
plt.xlabel("Age")
plt.ylabel("Sales")
sns.regplot(data=df,x="Age",y="Sales")
plt.show()



```
